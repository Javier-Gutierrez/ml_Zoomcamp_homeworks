```python
#Libraries 

import numpy as np
import pandas as pd
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction import DictVectorizer
from sklearn.tree import DecisionTreeRegressor
from sklearn.tree import export_text
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
```

```python
#Loading dataset

data = 'https://raw.githubusercontent.com/Javier-Gutierrez/machine-learning-zoomcamp/refs/heads/main/cohorts/2026/data/car_fuel_efficiency_2026.csv'
```

```python
#Reading data 

df = pd.read_csv(data)

#Showing first 5 rows
df.head()
```

```python
#EDA
df.nunique()
```

```python
#EDA

df.isnull().sum()
```

```python
df.info()
```

```python
#Filling missing values with 0

df['horsepower'] = df['horsepower'].fillna(0)
df['acceleration'] = df['acceleration'].fillna(0)
```

```python
df.isnull().sum()
```

```python
#Splitting the dataset

df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=1)
```

```python
#Reseting index

df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
df_full_train = df_full_train.reset_index(drop=True)
```

```python
#Y values
y_train = df_train.fuel_efficiency_mpg.values
y_val = df_val.fuel_efficiency_mpg.values
y_test = df_test.fuel_efficiency_mpg.values
```

```python
#Deleting the target from my datasets for avoiding problems

del df_train['fuel_efficiency_mpg']
del df_val['fuel_efficiency_mpg']
del df_test['fuel_efficiency_mpg']
del df_full_train['fuel_efficiency_mpg']
```

```python

#Making One-hot encoding of train dataset

#Train dataset
train_dicts = df_train.to_dict(orient='records')
#Genereting and instance for Vectorizer
dv_train = DictVectorizer(sparse=True)
X_train = dv_train.fit_transform(train_dicts)

#Validation dataset
val_dicts = df_val.to_dict(orient='records')
X_val = dv_train.transform(val_dicts)
```

```python
#Training decision tree

dt = DecisionTreeRegressor(max_depth=1)

dt.fit(X_train, y_train)
```

```python
#Q1 Looking for the split characteristic

print(export_text(dt, feature_names=dv_train.get_feature_names_out()))
```

```python
#Q2 Random forest

rf = RandomForestRegressor(n_estimators=10,  random_state=1, n_jobs=-1)

rf.fit(X_train, y_train)

#MSE
y_pred = rf.predict(X_val)
mse = mean_squared_error(y_val, y_pred)

#RMSE
rmse = np.sqrt(mse)

print(rmse)
```

```python
#Q3 Estimators

estimator = [10, 50, 100, 150]

for e in estimator:
    rf = RandomForestRegressor(n_estimators=e,  random_state=1, n_jobs=-1)

    rf.fit(X_train, y_train)

    #MSE
    y_pred = rf.predict(X_val)
    mse = mean_squared_error(y_val, y_pred)

    #RMSE
    rmse = np.sqrt(mse)

    print(rmse.round(3))
```

```python
#Q4 Max depth

max_depth = [10,15,20,25]
estimator = [10, 50, 100, 150]
columns = ['max_depth', 'estimator', 'rmse']
res = []

for depth in max_depth:
    for e in estimator:
        rf = RandomForestRegressor(n_estimators=e,  max_depth=depth, random_state=1, n_jobs=-1)

        rf.fit(X_train, y_train)

        #MSE
        y_pred = rf.predict(X_val)
        mse = mean_squared_error(y_val, y_pred)

        #RMSE
        rmse = np.sqrt(mse).round(3)
        res.append((depth,e,rmse))

df_results = pd.DataFrame(res, columns=columns)
df_results

        
```

```python
#Q4
print(df_results.loc[df_results['rmse'].idxmin()])
```

```python
#Q5
rf = RandomForestRegressor(n_estimators=10,  max_depth=20, random_state=1, n_jobs=-1)
rf.fit(X_train, y_train)

name_columns = dv_train.get_feature_names_out()
importances = rf.feature_importances_

df_importances = pd.DataFrame({'feature':name_columns, 'importance':importances})

print(df_importances.sort_values(by='importance', ascending=False))
```

```python
#Q6
features =list(dv_train.get_feature_names_out())
dtrain = xgb.DMatrix(X_train, label=y_train, feature_names=features)
dval = xgb.DMatrix(X_val, label=y_val, feature_names=features)

#watchlist
watchlist = [(dtrain, 'train'), (dval,'val')]

xgb_params = {
    'eta': 0.3, 
    'max_depth': 6,
    'min_child_weight': 1,
    
    'objective': 'reg:squarederror',
    'eval_metric': 'rmse',
    'nthread': 8,
    
    'seed': 1,
    'verbosity': 1,
}

model = xgb.train(xgb_params, dtrain, num_boost_round=100, evals=watchlist, verbose_eval=10)
```
