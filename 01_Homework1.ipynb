```python
#Q7 1.- Cars from ASIA

df_Asia = df[df['origin'] == 'Asia']

df_Asia
```

```python
#Q7 2.- Filtering by weight and model year
df_Asia_WY = df_Asia[['vehicle_weight','model_year']]
df_Asia_WY
```

```python
#Q7 3.- Selecting the first 7 val
df_7_values=df_Asia_WY.head(7)
df_7_values
```

```python
#Q7 4.-Get the underlying NmPy array. X
X=np.array(df_7_values)
X
```

```python
#Q7 5.- Matrix multiplication with X transpose

XTX = X.T.dot(X)
XTX
```

```python
#Q7 6.- Getting XTX inverted
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
#Checking for the identity matrix
XTX.dot(XTX_inv).round(1)
```

```python
#Q7 7.-Creating y values

y = np.array([1100,1300,800,900,1000,1100,1200],)
y.shape
```

```python
#Q7 8.- Multiplying XTX_INV dot XTX dot y

w = XTX_inv.dot(X.T).dot(y)
w
```

```python
#Q7 9. Sum of weights

w.sum()
```
